```python
# Mengimpor pustaka google_play_scraper untuk mengakses ulasan dan informasi aplikasi dari Google Play Store.
from google_play_scraper import app, reviews, Sort
import csv

# Daftar ID Aplikasi Travel
apps = {
    'Traveloka': 'com.traveloka.android',
    'Tiket.com': 'com.tiket.gits',
    'Agoda': 'com.agoda.mobile.consumer'
}

all_reviews_list = []


# Scraping
for app_name, app_id in apps.items():
    print(f"Sedang mengambil data untuk: {app_name}...")

    scrapreview, _ = reviews(
        app_id,
        lang='id',
        country='id',
        sort=Sort.MOST_RELEVANT,
        count=50000
    )

    # Tambahkan identitas nama aplikasi ke setiap ulasan
    for r in scrapreview:
        r['appName'] = app_name

    # Masukkan ke list besar
    all_reviews_list.extend(scrapreview)

# Membuat Dataframe
app_reviews_df = pd.DataFrame(all_reviews_list)
```

```python
app_reviews_df.to_csv('ulasan_travel_apps.csv', index=False, encoding='utf-8-sig')
```
